### STEP 0 - IMPORT RELEVANT PACKAGES

```python
import numpy as np
import os
from pathlib import Path
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

### STEP 1 - DOWNLOAD THE INITIAL DATA

```python
# Set the base directory path
base_dir = Path('/Users/jackdunning/Documents/Github/Capstone-Project-Imperial-College-London//data/initial_data')

# Dictionary to store all loaded data
initial_data = {}

# Load data for each function
for function_num in range(1, 9):
    function_name = f'function_{function_num}'
    function_dir = base_dir / function_name
    
    # Load inputs and outputs
    inputs_path = function_dir / 'initial_inputs.npy'
    outputs_path = function_dir / 'initial_outputs.npy'
    
    if inputs_path.exists() and outputs_path.exists():
        # Load data using np.load()
        inputs = np.load(inputs_path)
        outputs = np.load(outputs_path)
        
        initial_data[function_name] = {
            'inputs': inputs,
            'outputs': outputs
        }
    else:
        print(f"Warning: Files not found for {function_name}")
```

### STEP 2 - UNDERSTAND AND VISUALISE THE DATA

```python
# Helper function to visualize function data
def plot_function_data(function_name, inputs, outputs):
    """
    Plot inputs and outputs for a function.
    
    Parameters:
    - function_name: str, name of the function
    - inputs: numpy array, input data
    - outputs: numpy array, output data
    """
    n_dims = inputs.shape[1]
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    
    # Scatter plot (first two dimensions)
    scatter = axes[0].scatter(inputs[:, 0], inputs[:, 1], c=outputs, cmap='viridis', s=100)
    axes[0].set_xlabel('Input 1')
    axes[0].set_ylabel('Input 2')
    title_suffix = 'Input Space' if n_dims == 2 else 'First Two Dimensions'
    axes[0].set_title(f'{function_name} - {title_suffix}')
    axes[0].grid(True, alpha=0.3)
    plt.colorbar(scatter, ax=axes[0], label='Output')
    
    # Output distribution
    axes[1].hist(outputs, bins=min(10, len(outputs)), edgecolor='black')
    axes[1].set_xlabel('Output Value')
    axes[1].set_ylabel('Frequency')
    axes[1].set_title(f'{function_name} - Output Distribution')
    axes[1].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
```

```python
## Function 1

# Print data
function_name = 'function_1'
data = initial_data[function_name]
inputs = data['inputs']
outputs = data['outputs']

print(f"{function_name}:")
print(f"  Inputs shape: {inputs.shape}")
print(f"  Outputs shape: {outputs.shape}")
print(f"  Inputs:\n{inputs}")
print(f"  Outputs:\n{outputs}")

# Visualize
plot_function_data(function_name, inputs, outputs)
```
